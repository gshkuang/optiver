## Memory reduce function

```python
def reduce_mem_usage(df, verbose=0):
    """
    作用：遍历DataFrame数值列，依据取值范围压缩到更小的dtype以降低内存占用。
    注意：仅处理非object列；保证在所选dtype边界内不会溢出。
    返回：优化后的DataFrame。
    """

    # 记录优化前的内存占用（MB）
    start_mem = df.memory_usage().sum() / 1024**2

    for col in df.columns:
        col_type = df[col].dtype  # 当前列的数据类型

        # 仅处理非object类型（数值列），字符串/分类列跳过
        if col_type != object:
            # 计算该列的最小值与最大值，用于判断是否可安全缩到更小的整数/浮点范围
            c_min = df[col].min()
            c_max = df[col].max()
            if str(col_type)[:3] == "int":
                # 整型列：按取值范围从小到大尝试 int8→int16→int32→int64
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                elif c_min > np.iinfo(np.int64).min and c_max < np.iinfo(np.int64).max:
                    df[col] = df[col].astype(np.int64)
            else:
                # 浮点列：统一转换为 float32（在多数场景下内存/精度折中更优）
                if c_min > np.finfo(np.float16).min and c_max < np.finfo(np.float16).max:
                    df[col] = df[col].astype(np.float32)
                elif c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float32)

    if verbose:
        # verbose 日志输出（可选）：展示优化前后内存及降幅
        logger.info(f"Memory usage of dataframe is {start_mem:.2f} MB")
        end_mem = df.memory_usage().sum() / 1024**2
        logger.info(f"Memory usage after optimization is: {end_mem:.2f} MB")
        decrease = 100 * (start_mem - end_mem) / start_mem
        logger.info(f"Decreased by {decrease:.2f}%")

    return df
```

## Speed up triplet imbalance calculations

```python
from numba import njit, prange

@njit(parallel=True)
# compute_triplet_imbalance：并行计算三元组不平衡度，输入为二维数组与三元组索引，输出不平衡度矩阵
def compute_triplet_imbalance(df_values, comb_indices):
    num_rows = df_values.shape[0]
    num_combinations = len(comb_indices)
    imbalance_features = np.empty((num_rows, num_combinations))

    for i in prange(num_combinations):
        a, b, c = comb_indices[i]
        for j in range(num_rows):
            max_val = max(df_values[j, a], df_values[j, b], df_values[j, c])
            min_val = min(df_values[j, a], df_values[j, b], df_values[j, c])
            mid_val = df_values[j, a] + df_values[j, b] + df_values[j, c] - min_val - max_val
            if mid_val == min_val:  # Prevent division by zero
                imbalance_features[j, i] = np.nan
            else:
                imbalance_features[j, i] = (max_val - mid_val) / (mid_val - min_val)

    return imbalance_features


# calculate_triplet_imbalance_numba：封装DataFrame列的三元组不平衡度计算，返回命名完备的特征DataFrame
def calculate_triplet_imbalance_numba(price, df):
    # Convert DataFrame to numpy array for Numba compatibility
    df_values = df[price].values
    comb_indices = [(price.index(a), price.index(b), price.index(c)) for a, b, c in combinations(price, 3)]

    # Calculate the triplet imbalance
    features_array = compute_triplet_imbalance(df_values, comb_indices)

    # Create a DataFrame from the results
    columns = [f"{a}_{b}_{c}_imb2" for a, b, c in combinations(price, 3)]
    features = pd.DataFrame(features_array, columns=columns)

    return features
```

## Feature groups

```python
# generate imbalance features
def imbalance_features(df):
    # 价格与数量的基础字段集合，用于生成组合特征
    prices = ["reference_price", "far_price", "near_price", "ask_price", "bid_price", "wap"]
    sizes = ["matched_size", "bid_size", "ask_size", "imbalance_size"]

    # V1：基础价量/失衡特征
    df["volume"] = df.eval("ask_size + bid_size")  # 总挂单规模，刻画流动性深度
    df["mid_price"] = df.eval("(ask_price + bid_price) / 2")  # 中间价，价差的中心位置
    df["liquidity_imbalance"] = df.eval("(bid_size-ask_size)/(bid_size+ask_size)")  # 委托量偏向度，[-1,1]，>0 买方更强
    df["matched_imbalance"] = df.eval("(imbalance_size-matched_size)/(matched_size+imbalance_size)")  # 成交与失衡量的相对偏差，衡量买卖压力
    df["size_imbalance"] = df.eval("bid_size / ask_size")  # 尺度比，>1 买方深度更大
    
    for c in combinations(prices, 2):
        # 两两价格相对不平衡度：尺度化(差/和)，抵消量纲与大小级差异
        df[f"{c[0]}_{c[1]}_imb"] = df.eval(f"({c[0]} - {c[1]})/({c[0]} + {c[1]})")

    for c in [['ask_price', 'bid_price', 'wap', 'reference_price'], sizes]:
        # 三元组不平衡度：使用 (max-mid)/(mid-min) 衡量极端偏向，Numba并行加速
        triplet_feature = calculate_triplet_imbalance_numba(c, df)
        df[triplet_feature.columns] = triplet_feature.values
        
    # V2：交互、动量与派生特征
    df["stock_weights"] = df["stock_id"].map(weights)  # 股票权重，用于加权成交价（经验权重）
    df["weighted_wap"] = df["stock_weights"] * df["wap"]  # 权重加权WAP，提升重要股票的影响
    df['wap_momentum'] = df.groupby('stock_id')['weighted_wap'].pct_change(periods=6)  # 6步WAP动量，短期趋势
    df["imbalance_momentum"] = df.groupby(['stock_id'])['imbalance_size'].diff(periods=1) / df['matched_size']  # 供需变化强度：失衡变化/成交量
    df["price_spread"] = df["ask_price"] - df["bid_price"]  # 买卖价差，交易成本与波动空间
    df["spread_intensity"] = df.groupby(['stock_id'])['price_spread'].diff()  # 价差一阶变化，收窄/扩张速度
    df['price_pressure'] = df['imbalance_size'] * (df['ask_price'] - df['bid_price'])  # 价格压力：失衡量×价差
    df['market_urgency'] = df['price_spread'] * df['liquidity_imbalance']  # 市场紧迫度：价差与流动性偏向的交互
    df['depth_pressure'] = (df['ask_size'] - df['bid_size']) * (df['far_price'] - df['near_price'])  # 深度压力：深度差×远近价差
    df['spread_depth_ratio'] = (df['ask_price'] - df['bid_price']) / (df['bid_size'] + df['ask_size'])  # 单位深度的价差
    df['mid_price_movement'] = df['mid_price'].diff(periods=5).apply(lambda x: 1 if x > 0 else (-1 if x < 0 else 0))  # 中间价5步方向符号
    df['micro_price'] = ((df['bid_price'] * df['ask_size']) + (df['ask_price'] * df['bid_size'])) / (df['bid_size'] + df['ask_size'])  # Micro Price：订单量加权微观价格
    df['relative_spread'] = (df['ask_price'] - df['bid_price']) / df['wap']  # 相对价差：用成交均价归一
    
    for func in ["mean", "std", "skew", "kurt"]:
        # 行级价格/尺寸集合的统计汇总：均值/标准差/偏度/峰度
        df[f"all_prices_{func}"] = df[prices].agg(func, axis=1)
        df[f"all_sizes_{func}"] = df[sizes].agg(func, axis=1)
        
    # V3：时序派生特征（滞后与差分）
    for col in ['matched_size', 'imbalance_size', 'reference_price', 'imbalance_buy_sell_flag']:
        for window in [1, 2, 3, 5, 10]:
            df[f"{col}_shift_{window}"] = df.groupby('stock_id')[col].shift(window)  # t-window 的历史值
            df[f"{col}_ret_{window}"] = df.groupby('stock_id')[col].pct_change(window)  # 相对变化率（收益率）
            
    for col in ['ask_price', 'bid_price', 'ask_size', 'bid_size',
                'wap', 'near_price', 'far_price']:
        for window in [1, 2, 3, 5, 10]:
            df[f"{col}_diff_{window}"] = df.groupby("stock_id")[col].diff(window)  # 短期变化幅度（差分）

    return df.replace([np.inf, -np.inf], 0)  # 将无穷值替换为0以稳定训练

# generate time & stock features
def other_features(df):
    # 时间维度衍生：date_id 为交易日编号（1开始）；seconds_in_bucket 为当日的秒级时间（每分钟一个桶）
    df["dow"] = df["date_id"] % 5    # 周内日索引（1~5），刻画每周结构性差异
    df["dom"] = df["date_id"] % 20   # 月内日索引（近似1~20），刻画月份内效应
    df["seconds"] = df["seconds_in_bucket"] % 60  # 当分钟的秒（0~59）
    df["minute"] = df["seconds_in_bucket"] // 60   # 当日的分钟序号（0~599）

    # 股票层面的全局统计特征：在训练集上按 stock_id 预先统计的量，推断时按映射填充
    for key, value in global_stock_id_feats.items():
        df[f"global_{key}"] = df["stock_id"].map(value.to_dict())

    return df

# generate all features
def generate_all_features(df):
    # 统一的特征生成入口：过滤无关列→价量/失衡特征→时间/股票全局特征
    cols = [c for c in df.columns if c not in ["row_id", "time_id", "target"]]
    df = df[cols]
    df = imbalance_features(df)
    df = other_features(df)
    gc.collect()
    
    # 最终用于训练/推断的特征列（排除标识/目标/时间ID）
    feature_name = [i for i in df.columns if i not in ["row_id", "target", "time_id", "date_id"]]
    
    return df[feature_name]
```

> Because this is a time series dataset, we cannot use random KFold to partition the data, which will lead to data leakage (the model will have difficulty converging). Therefore, we strictly ensure that a part of the test set is not used, and divide all *training* data into (train, valid, test) in time series, valid is used as the basis for train iteration, and the optimal parameters are used on train.

```python
if is_train:
    feature_name = list(df_train_feats.columns)
    lgb_params = {
        "objective" : "mae",          # 损失为绝对误差（Mean Absolute Error），与比赛指标一致
        "n_estimators" : 3000,      # 最大迭代轮数（树数），配合早停确定有效树
        "num_leaves" : 128,        # 叶子数，控制树的复杂度与拟合能力
        "subsample" : 0.6,         # 行采样比例，提升泛化、降低过拟合
        "colsample_bytree" : 0.6,  # 列采样比例，每棵树的特征子集
        "learning_rate" : 0.05,    # 学习率，降低每棵树对总体的影响幅度
        "n_jobs" : 4,              # 并行线程数
        "device" : "gpu",        # 使用GPU加速（如可用）
        "verbosity": -1,          # 静默日志
        "importance_type" : "gain", # 特征重要性采用增益
    }

    print(f"Feature length = {len(feature_name)}")

    offline_split = df_train['date_id']>(split_day - 45)  # 将最后45天作为验证窗口
    df_offline_train = df_train_feats[~offline_split]
    df_offline_valid = df_train_feats[offline_split]
    df_offline_train_target = df_train['target'][~offline_split]
    df_offline_valid_target = df_train['target'][offline_split]

    print("Valid Model Trainning.")
    lgb_model = lgb.LGBMRegressor(**lgb_params)
    lgb_model.fit(
        df_offline_train[feature_name],
        df_offline_train_target,
        eval_set=[(df_offline_valid[feature_name], df_offline_valid_target)],
        callbacks=[
            lgb.callback.early_stopping(stopping_rounds=100),   # 早停：验证集100轮未提升则停止
            lgb.callback.log_evaluation(period=100),            # 每100步打印一次验证日志
        ],
    )

    del df_offline_train, df_offline_valid, df_offline_train_target, df_offline_valid_target
    gc.collect()

    # infer
    df_train_target = df_train["target"]
    print("Infer Model Trainning.")
    infer_params = lgb_params.copy()
    infer_params["n_estimators"] = int(1.2 * lgb_model.best_iteration_)  # 用最佳迭代数的1.2倍在全量train上重训用于推断
    infer_lgb_model = lgb.LGBMRegressor(**infer_params)
    infer_lgb_model.fit(df_train_feats[feature_name], df_train_target)

    if is_offline:   
        # offline predictions
        df_valid_target = df_valid["target"]
        offline_predictions = infer_lgb_model.predict(df_valid_feats[feature_name])
        offline_score = mean_absolute_error(offline_predictions, df_valid_target)  # 线下MAE评估
        print(f"Offline Score {np.round(offline_score, 4)}")
```

## 时序划分与训练策略
- 严格避免信息泄露：以时间顺序划分 train/valid/test
- 先在 valid 上做早停与参数选择，再用最佳迭代数的 1.2 倍在全量 train 上拟合用于推断

```python
# zero_sum：对预测序列施加零和约束，按体量的不确定性（sqrt(volume)）做加权调整
def zero_sum(prices, volumes):
    std_error = np.sqrt(volumes)  # 体量的不确定性尺度，用于权重
    step = np.sum(prices)/np.sum(std_error)  # 使加权和为零的步进
    out = prices-std_error*step  # 生成零和约束后的预测
    
    return out

if is_infer:
    import optiver2023
    env = optiver2023.make_env()
    iter_test = env.iter_test()
    counter = 0
    y_min, y_max = -64, 64
    qps, predictions = [], []
    cache = pd.DataFrame()
    for (test, revealed_targets, sample_prediction) in iter_test:
        now_time = time.time()
        cache = pd.concat([cache, test], ignore_index=True, axis=0)
        if counter > 0:
            cache = cache.groupby(['stock_id']).tail(21).sort_values(by=['date_id', 'seconds_in_bucket', 'stock_id']).reset_index(drop=True)
        feat = generate_all_features(cache)[-len(test):]
        lgb_prediction = infer_lgb_model.predict(feat)
#         lgb_prediction = zero_sum(lgb_prediction, test['bid_size'] + test['ask_size'])
        lgb_prediction = lgb_prediction - np.mean(lgb_prediction)
        clipped_predictions = np.clip(lgb_prediction, y_min, y_max)
        sample_prediction['target'] = clipped_predictions
        env.predict(sample_prediction)
        counter += 1
        qps.append(time.time() - now_time)
        if counter % 10 == 0:
            print(counter, 'qps:', np.mean(qps))
           
    time_cost = 1.146 * np.mean(qps)
    print(f"The code will take approximately {np.round(time_cost, 4)} hours to reason about")
```
